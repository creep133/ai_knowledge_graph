```python
from typing import Dict

# 'user'는 문자열(str) 키와 문자열(str) 값을 가진 딕셔너리(Dict)여야 함
def greet_user(user: Dict[str, str]) -> str:
    return f"Hello, {user['name']}! You are {user['age']} years old {user['sex']}."

# 사용 예
user_data = {"name": "Tanjiro", "age": "14", "sex": "male"}
print(greet_user(user_data))
```

```python
from typing import Union

def format_id(item_id: Union[int, str]) -> str:
    return f"Item-ID-{item_id}"

print(format_id(101))       # "Item-ID-101"
print(format_id("A-500"))   # "Item-ID-A-500"
```

```python
def add_numbers(a: int, b: int) -> int:
    return a + b

# 1. 이건 잘 됨
print(add_numbers(10, 20))

# 2. '실행'하기도 전에 VS Code가 바로 경고!
print(add_numbers("안녕", 5))
```

```python
from pydantic import BaseModel

# "경비원(BaseModel)님, 오늘 파티 규칙은 이겁니다:"
class User(BaseModel):
    # 규칙 1: 'name'이라는 이름표가 있어야 하고, 반드시 '문자열(str)'이어야 함.
    name: str 
    
    # 규칙 2: 'age'라는 이름표가 있어야 하고, 반드시 '정수(int)'여야 함.
    age: int
```

```python
# 손님 1 도착
good_data = {"name": "Tanjiro", "age": 15}

# 경비원에게 검사 요청
user1 = User(**good_data) 

# 경비원 💂: "OK. 'name'은 문자열, 'age'는 정수. 명단과 일치합니다. 통과!"
# 결과: user1이라는 'User' 객체가 성공적으로 생성됩니다.
```
